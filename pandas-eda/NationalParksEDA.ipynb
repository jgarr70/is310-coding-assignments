```python
import pandas as pd

np_data = pd.read_csv("https://raw.githubusercontent.com/melaniewalsh/responsible-datasets-in-context/main/datasets/national-parks/US-National-Parks_RecreationVisits_1979-2024.csv")
```

```python
print(np_data.columns)

avg_state_visits = np_data.groupby("State")['RecreationVisits'].mean()
# print(avg_state_visits)

avg_park_visits = np_data.groupby("ParkName")['RecreationVisits'].mean()
# print(avg_park_visits)

distinct_parks = np_data.groupby("State")["ParkName"].nunique()
print(distinct_parks)
```
